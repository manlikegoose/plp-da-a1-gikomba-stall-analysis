# Gikomba Market Stall Analysis

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2



# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

  "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

  "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

  "Sukuma": {"qty": 9,"reorder": 15, "cost": 30, "price": 50},

  "Potatoes":{"qty": 55, "reorder": 30, "cost": 50, "price": 75},

  "Avocado":{"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}
print(stock
      )
```

```python

def total(sales):
  total_sale = sum(sales)
  average = total_sale/len(sales)
  max_sales=max(sales)
  min_sales = min(sales)

  pos = sales.index(max(sales))
  pos1 = sales.index(min(sales))

  week_number = (pos // 7) + 1




  print(f"The total sales is : {total_sale}")
  print(f"The average sales for the two weeks is : {average :,.2f}")
  print(f"Best day: {days[pos]} (position {pos}) and the maximum sale for the weeks is KES {max_sales}")
  print(f"worse day: {days[pos1]} (position {pos1}) and the minimum sale for the weeks is KES {min_sales}")
  print(f"There are {week_number} weeks")
total(daily_sales)

# avarage
```

```python
def week_totals(sales):
  week1 =sum( sales[0:7])
  week2 =sum( sales[7:])

  return week1,week2

w1_total,w2_total = week_totals(daily_sales)

growth = ((w2_total-w1_total)/w1_total)*100

print(f"Growth is {growth:,.2f}%")
print(f"week 1 total was: KES {w1_total}")
print(f"week 2 total was: KES { w2_total}")
```

```python
def classify_day(amount, average):
    pct_avg = ((amount - average) / average) * 100

    if pct_avg > 20:
        return "Strong"
    elif pct_avg < -20:
        return "Weak"
    else:
        return "Normal"


counts = {"Strong": 0, "Weak": 0, "Normal": 0}

average = sum(daily_sales) / len(daily_sales)

for amount in daily_sales:
    category = classify_day(amount, average)
    counts[category] += 1

print("Average:", round(average, 2))

for amount in daily_sales:
    category = classify_day(amount, average)
    print(amount, "=", category)

print("Summary:", counts)
```

```python
def diff_weekday_weekend (sales):
  weekday =sales[0:5] + sales[7:12]
  weekend = sales[5:7] + sales[12:]

  avg_weekday = sum(weekday) /len(weekday)
  avg_weekend = sum(weekend) /len(weekend)

  pct_diff = ((avg_weekend - avg_weekday)/ avg_weekday)*100


  print(f"The percentage difference is :{pct_diff:,.2f}%")
  print(f"weekday average sale is KES {avg_weekday}")
  print(f"weekend average sale is KES {avg_weekend}")

diff_weekday_weekend (daily_sales)
```

```python
def restock_check(items):
    to_reorder = []

    for name, details in items.items():

        qty = details["qty"]
        reorder = details["reorder"]
        cost = details["cost"]
        price = details["price"]


        gross_margin = ((price - cost) / price) * 100


        if qty <= reorder:
            status = "Reorder"
            to_reorder.append(name)
        else:
            status = "OK"

        print(f"{name:<10} Qty: {qty:<3} Margin: {gross_margin:.2f}% Status: {status}")

    return to_reorder


result = restock_check(stock)

print("\nProducts to reorder:", result)
```

```python
def stall_performance_report(sales, days, stock):
    """Generate a formatted performance report for Wanjiku's stall."""


    total_sales = sum(sales)
    average_sales = total_sales / len(sales)

    highest_sales = max(sales)
    lowest_sales = min(sales)

    best_pos = sales.index(highest_sales)
    worst_pos = sales.index(lowest_sales)

    best_week = (best_pos // 7) + 1
    worst_week = (worst_pos // 7) + 1


    week1, week2 = week_totals(sales)
    growth = ((week2 - week1) / week1) * 100


    counts = {"Strong": 0, "Normal": 0, "Weak": 0}

    for amount in sales:
        category = classify_day(amount, average_sales)
        counts[category] += 1

    weekdays = sales[0:5] + sales[7:12]
    weekends = sales[5:7] + sales[12:14]

    weekday_average = sum(weekdays) / len(weekdays)
    weekend_average = sum(weekends) / len(weekends)

    percentage_difference = (
        (weekend_average - weekday_average) / weekday_average
    ) * 100


    to_reorder = []


    print("=" * 55)
    print("             STALL PERFORMANCE REPORT")
    print("                  GIKOMBA MARKET")
    print("=" * 55)

    print("\nOVERALL SALES PERFORMANCE")
    print("-" * 55)

    print(f"Total sales:          KES {total_sales:>12,.2f}")
    print(f"Average daily sales:  KES {average_sales:>12,.2f}")
    print(f"Highest sales:        KES {highest_sales:>12,.2f}")
    print(f"Lowest sales:         KES {lowest_sales:>12,.2f}")

    print(
        f"Best day: {days[best_pos]}, Week {best_week}"
    )
    print(
        f"Worst day: {days[worst_pos]}, Week {worst_week}"
    )

    print("\nWEEKLY PERFORMANCE")
    print("-" * 55)

    print(f"Week 1 sales:         KES {week1:>12,.2f}")
    print(f"Week 2 sales:         KES {week2:>12,.2f}")
    print(f"Week-on-week growth:  {growth:>12.2f}%")

    print("\nDAILY SALES CLASSIFICATION")
    print("-" * 55)

    for category, count in counts.items():
        print(f"{category:<15} {count:>5} days")

    print("\nWEEKDAY VS WEEKEND")
    print("-" * 55)

    print(f"Weekday average:      KES {weekday_average:>12,.2f}")
    print(f"Weekend average:      KES {weekend_average:>12,.2f}")
    print(f"Percentage difference: {percentage_difference:>11.2f}%")

    print("\nSTOCK PERFORMANCE")
    print("-" * 55)

    print(f"{'Product':<12}{'Qty':>6}{'Margin':>12}{'Status':>12}")

    for name, details in stock.items():

        qty = details["qty"]
        reorder = details["reorder"]
        cost = details["cost"]
        price = details["price"]

        gross_margin = ((price - cost) / price) * 100

        if qty <= reorder:
            status = "REORDER"
            to_reorder.append(name)
        else:
            status = "OK"

        print(
            f"{name:<12}{qty:>6}"
            f"{gross_margin:>11.2f}%{status:>12}"
        )

    print("\nRESTOCK RECOMMENDATIONS")
    print("-" * 55)

    if to_reorder:
        for product in to_reorder:
            print(f"- {product}")
    else:
        print("All products have sufficient stock.")

    print("=" * 55)
    print("                 END OF REPORT")
    print("=" * 55)



stall_performance_report(daily_sales, days, stock)
```
